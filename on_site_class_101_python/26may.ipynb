## ✅ What is Recursion in Python?

Recursion means a function calling itself to solve smaller parts of a problem.

It keeps calling itself until it reaches a base case, which stops the recursion.

⚠️ Important:
Always include a base case to avoid infinite loops.

Be careful with large numbers, recursion can hit the Python recursion limit.

```python
def factorial(n):
    
    if n <0:
        return None
    if n < 2:
        return 1
    else:
        result = 1
        
        for i in range(2,n+1):
            result *= i
        return result
 
    
output = factorial(5)  # Output: 120
print("Factorial of 5 is:", output)  # Output: Factorial of 5 is: 120

    
print(factorial(5))  # Output: 120

    
```

```python
def factorial(n):
    
    if n < 0:
        return None
    if n < 2:
        return 1
    
    else:  
        return n * factorial(n - 1)
    
    
print(factorial(3))  # Output: 120
```

🔢 Fibonacci Formula (Recursive Definition):

For position n, the Fibonacci number is:

F(n) = F(n-1) + F(n-2)

```python
def fib(n):
    if n <=0:
        return 0
    if n == 1:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)  
    
    ### first step  fib(4) + fib(3)
    ## once again fib function call fib(4) = fib(3) + fib(2)=> = 2 + 1 = 3
    ## once again fib function call fib(3) = fib(2) + fib(1)=> 1+1 = 2
    ## once again fib function call fib(2) = fib(1) + fib(0) = 1
    ## once again fib function call fib(1) = 1
    ## once again fib function call fib(0) = 0
    ## put all values from f(0) to f(1) 
print(fib(5))  # Output: 5
```

```python
def fib(n):
    if n <= 0:
        return 0      # Base case: fib(0) = 0
    if n == 1:
        return 1      # Base case: fib(1) = 1
    else:
        # Recursive case:
        # fib(n) = fib(n-1) + fib(n-2)
        return fib(n - 1) + fib(n - 2)

# Example: fib(5)

# fib(5) = fib(4) + fib(3)
# fib(4) = fib(3) + fib(2)
# fib(3) = fib(2) + fib(1)
# fib(2) = fib(1) + fib(0)

# Now, substitute base values:
# fib(1) = 1
# fib(0) = 0

# Calculate step-by-step:
# fib(2) = 1 + 0 = 1
# fib(3) = fib(2) + fib(1) = 1 + 1 = 2
# fib(4) = fib(3) + fib(2) = 2 + 1 = 3
# fib(5) = fib(4) + fib(3) = 3 + 2 = 5

print(fib(5))  # Output: 5
```

## What is a Lambda?

A lambda is a super short way to write a function.

It’s like saying, “Hey, here’s a quick job for you — do this and give me the answer!” without making a big long list of instructions.

Why do we need it?
Sometimes, you want to do a quick little job only once and you don’t want to write a full big function.

It keeps your code simple and short.

🐍 What is **kwargs?
💡 kwargs stands for:
Keyword Arguments

✅ Simple Explanation:
You use **kwargs when you want your function to accept many named (keyword) arguments.

It packs those extra keyword arguments into a dictionary.

```python
def greet(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

greet(name="Alice", age=7, hobby="drawing")
```

```python
## key words Arguments
### kwargs

def simple_func(**kwargs):
    for key, values in kwargs.items():
        print(f"{key} is {values}")
    # print(kwargs.items())

simple_func(name = "Danish", age = 34)
```
