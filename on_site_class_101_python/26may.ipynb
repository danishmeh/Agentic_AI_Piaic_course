What is the sum of the elements in the second column?

```python
matrix = [
  [5, 10, 15],
  [20, 25, 30],
  [35, 40, 45]
]



total = 0
# for row in matrix:
##row[1] means get the second item in each row
column = [row[1] for row in matrix]

for col in column:
    total += col
print("Column:", column)  # Output: [5, 20, 35]
print("Total of column 1:", total)  # Output: 60
```

```python
# Write code to transpose this 2x3 matrix (swap rows and columns):
matrix = [
  [1, 2, 3],
  [4, 5, 6]
]
transposed_matrix = []

for i in range(len(matrix[0])):  # Iterate over columns
    new_row = []
    for j in range(len(matrix)):  # Iterate over rows
        new_row.append(matrix[j][i])
    transposed_matrix.append(new_row)
        


print("Transposed matrix:", transposed_matrix)  # Output: [[1, 4], [2, 5], [3, 6]]
```
