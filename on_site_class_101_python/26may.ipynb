```python
# Write code to transpose this 2x3 matrix (swap rows and columns):
matrix = [
  [1, 2, 3],
  [4, 5, 6]
]
transposed_matrix = []

for i in range(len(matrix[0])):  # Iterate over columns
    new_row = []
    for j in range(len(matrix)):  # Iterate over rows
        new_row.append(matrix[j][i])
    transposed_matrix.append(new_row)
        


print("Transposed matrix:", transposed_matrix)  # Output: [[1, 4], [2, 5], [3, 6]]
```

➕ What does this mean?

```python
def add(a, b):
    result =  a + b  # Using return to give output
    return result  # This will return the sum of a and b

first_number = int(input("Enter the first number: "))
second_number = int(input("Enter the second number: "))
add = add(first_number, second_number)  # Output: The sum is: <sum of the two numbers>


print("Output:", add)  # Output: <sum of the two numbers>

def multiply(a, b):
    result = a * b  # Using return to give output
    return result  # This will return the product of a and b

last_output = multiply(add, second_number)
print("Output of multiply function:", last_output)  # Output: <product of the two numbers>
```

```python
def add(a, b):
    result = a + b  # Using return to give output
    return result

first_number = int(input("Enter the first number: "))
second_number = int(input("Enter the second number: "))
sum_result = add(first_number, second_number)  # Get the sum

def mul(add_result, b):
    output = add_result * b  # Multiply the sum with another number
    return output

# Call the mul function
final_output = mul(sum_result, second_number)

print("Output:", final_output)
```

```python
'''
b=5 means b has a default value of 5.

So, if you call add_numbers(3), since you only gave one argument (a=3), Python uses the default value for b, which is 5. So it calculates 3 + 5 = 8.

But if you call add_numbers(3, 7), you provide a value for b (7), so Python uses that instead of the default.

'''


def add_numbers(a, b=5):
    return a + b

print(add_numbers(3))
print(add_numbers(3, 7))
```
